### Notation, Explainability Methods, and Mask-Based Metrics

| Symbol | Name | Meaning | Type | Mathematical Formula | Python Formula (conceptual) |
|------|------|---------|------|----------------------|-----------------------------|
| $H$ | Saliency map | Continuous relevance heatmap produced by an XAI method | Real-valued matrix | $H = \{ H_{i,j} \mid H_{i,j} \ge 0 \}$ | `H` |
| $M$ | Lesion mask | Ground-truth binary mask of lesion region | Binary matrix | $M_{i,j} \in \{0,1\}$ | `M` |
| $\hat{M}$ | Non-lesion mask | Complement of lesion mask | Binary matrix | $\hat{M} = 1 - M$ | `1 - M` |
| $\tilde{H}$ | Thresholded saliency map | Binary saliency map derived from $H$ | Binary matrix | $\tilde{H}_{i,j} = \mathbb{1}[H_{i,j} \ge \tau]$ | `H >= tau` |
| $L$ | Lesion area | Number of lesion pixels | Scalar | $L = \sum_{i,j} M_{i,j}$ | `np.sum(M)` |
| $B$ | Background area | Number of background pixels | Scalar | $B = \sum_{i,j} \hat{M}_{i,j}$ | `np.sum(1 - M)` |
| $H^{GradCAM}$ | Grad-CAM heatmap | Class-discriminative saliency from CNN activations | Real-valued matrix | $H^{GradCAM} = \mathrm{ReLU}\!\left(\sum_k \alpha_k A^k\right)$ | `np.maximum(0, sum(alpha[k] * A[k] for k in channels))` |
| $\alpha_k$ | Grad-CAM weight | Importance of feature map $k$ for class $c$ | Scalar | $\alpha_k = \frac{1}{Z} \sum_{i,j} \frac{\partial y^c}{\partial A^k_{i,j}}$ | `grads[k].mean()` |
| $H^{LIME}$ | LIME saliency map | Pixel relevance from a local surrogate model | Real-valued matrix | $H^{LIME} = \sum_{s \in \mathcal{S}} w_s \mathbb{1}_s$ | `sum(w[s] * superpixel_mask[s] for s in S)` |
| $w_s$ | LIME weight | Contribution of superpixel $s$ | Scalar | $\hat{f}(x) = \sum_s w_s z_s$ | `lime_model.coef_` |
| $H^{SHAP}$ | SHAP saliency map | Pixel-wise Shapley value attribution map | Real-valued matrix | $H^{SHAP}_{i,j} = \phi_{i,j}$ | `np.abs(shap_values).sum(axis=0)` |
| $ \phi_{i,j}$ | SHAP value | Shapley value of pixel $(i,j)$ | Scalar | $\phi_i=\sum_{S\subseteq F\setminus\{i\}}\frac{\mathrm{card}(S)!\,(\mathrm{card}(F)-\mathrm{card}(S)-1)!}{\mathrm{card}(F)!}\bigl(f(S\cup\{i\})-f(S)\bigr)$ | `shap_values[i, j]` |
| $SIR$ | Saliency-Inside Ratio | Fraction of total saliency inside lesion | Scalar | $\frac{\sum_{i,j} H_{i,j} M_{i,j}}{\sum_{i,j} H_{i,j}}$ | `np.sum(H * M) / np.sum(H)` |
| $RD_{Lesion}$ | Lesion relevance density | Mean saliency per lesion pixel | Scalar | $\frac{\sum_{i,j} H_{i,j} M_{i,j}}{L}$ | `np.sum(H * M) / np.sum(M)` |
| $RD_{NoLesion}$ | Background relevance density | Mean saliency per background pixel | Scalar | $\frac{\sum_{i,j} H_{i,j} \hat{M}_{i,j}}{B}$ | `np.sum(H * (1 - M)) / np.sum(1 - M)` |
| $TIxAI$ | Trustworthiness Index for XAI | Difference between lesion and background relevance densities | Scalar | $TIxAI = RD_{Lesion} - RD_{NoLesion}$ | `(np.sum(H * M) / np.sum(M)) - (np.sum(H * (1 - M)) / np.sum(1 - M))` |
| $IoU$ | Intersection over Union | Spatial overlap between saliency and lesion | Scalar | $\frac{\sum_{i,j}\tilde{H}_{i,j}M_{i,j}}{\sum_{i,j}\left(\tilde{H}_{i,j}+M_{i,j}-\tilde{H}_{i,j}M_{i,j}\right)}$ | `np.sum(H_tilde & M) / np.sum(H_tilde + M - (H_tilde & M)) `|
| $Dice$ | Dice coefficient | Normalized spatial overlap | Scalar | $\frac{2 \sum_{i,j} \tilde{H}_{i,j} M_{i,j}} {\sum_{i,j} \tilde{H}_{i,j} + \sum_{i,j} M_{i,j}}$ | `2 * np.sum(H_tilde & M) / (np.sum(H_tilde) + np.sum(M))` |


```python
import numpy as np


def threshold_heatmap(H: np.ndarray, method: str = "topk", k: float = 0.2, tau: float = 0.5) -> np.ndarray:
    """
    Convert a continuous heatmap H into a binary map H_tilde.

    method:
      - "topk": keeps the top k fraction of pixels (k in (0,1])
      - "tau":  keeps pixels with H >= tau  (tau in [0,1] if H is normalized)

    Returns:
      H_tilde: bool array of same shape as H
    """
    H = np.asarray(H, dtype=float)

    if method == "topk":
        if not (0 < k <= 1):
            raise ValueError("k must be in (0, 1].")
        flat = H.ravel()
        # kth largest threshold
        idx = int(np.ceil((1 - k) * flat.size))
        thr = np.partition(flat, idx)[idx]
        return H >= thr

    if method == "tau":
        return H >= float(tau)

    raise ValueError("method must be 'topk' or 'tau'.")


def sir(H: np.ndarray, M: np.ndarray, eps: float = 1e-12) -> float:
    """
    Saliency-Inside Ratio (SIR):
        SIR = sum(H * M) / sum(H)
    H: continuous saliency map (non-negative recommended)
    M: binary lesion mask (0/1 or bool)
    """
    H = np.asarray(H, dtype=float)
    M = np.asarray(M).astype(bool)

    num = float(np.sum(H[M]))
    den = float(np.sum(H)) + eps
    return num / den


def tixai(H: np.ndarray, M: np.ndarray, eps: float = 1e-12) -> float:
    """
    TIxAI (as in Ieracitano et al. 2025):
        RD_lesion    = sum(H * M)   / sum(M)
        RD_nolesion  = sum(H * M^)  / sum(M^)
        TIxAI        = RD_lesion - RD_nolesion

    H: continuous saliency map
    M: binary lesion mask
    """
    H = np.asarray(H, dtype=float)
    M = np.asarray(M).astype(bool)
    M_hat = ~M

    L = float(np.sum(M)) + eps          # lesion area
    B = float(np.sum(M_hat)) + eps      # background area

    rd_lesion = float(np.sum(H[M])) / L
    rd_nolesion = float(np.sum(H[M_hat])) / B
    return rd_lesion - rd_nolesion


def iou(H_tilde: np.ndarray, M: np.ndarray, eps: float = 1e-12) -> float:
    """
    Intersection over Union (IoU) between binary maps:
        IoU = |H_tilde ∩ M| / |H_tilde ∪ M|
    """
    H_tilde = np.asarray(H_tilde).astype(bool)
    M = np.asarray(M).astype(bool)

    inter = float(np.sum(H_tilde & M))
    union = float(np.sum(H_tilde | M)) + eps
    return inter / union


def dice(H_tilde: np.ndarray, M: np.ndarray, eps: float = 1e-12) -> float:
    """
    Dice coefficient between binary maps:
        Dice = 2|H_tilde ∩ M| / (|H_tilde| + |M|)
    """
    H_tilde = np.asarray(H_tilde).astype(bool)
    M = np.asarray(M).astype(bool)

    inter = float(np.sum(H_tilde & M))
    size_sum = float(np.sum(H_tilde) + np.sum(M)) + eps
    return (2.0 * inter) / size_sum


# ---------- Example usage ----------
# H: continuous heatmap (e.g., Grad-CAM), shape (H,W)
# M: lesion mask, same shape (H,W), values 0/1 or bool

# H_tilde from top 20% most salient pixels:
# H_tilde = threshold_heatmap(H, method="topk", k=0.2)

# Or threshold by fixed tau (if H is normalized to [0,1]):
# H_tilde = threshold_heatmap(H, method="tau", tau=0.5)

# Metrics:
# SIR_val = sir(H, M)
# TIxAI_val = tixai(H, M)
# IoU_val = iou(H_tilde, M)
# Dice_val = dice(H_tilde, M)
```
